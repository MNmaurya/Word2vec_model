### Sample Text

```python
corpus = """The king is strong.
The queen is wise.
The prince is young."""
```

### Importing necessary libraraies

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
import re
text = re.sub(r"[^\w\s]", "", corpus)
```

```python
words=wordpunct_tokenize(text.lower())
print(words)
```

```python
print(len(words))
```

```python
vocab=sorted(set(words))  #Set function removes all the duplicates and sorted makes the consistency every time we generate output. This is helpful for learning and debugging.
print(vocab)
```

```python
len(vocab)
```

```python
word_to_index = {}  #dict 

for index, word in enumerate(vocab):
    word_to_index[word] = index

print(word_to_index)
```

```python
window_size=3

window=[]
for i in range(len(words)- window_size+1):
    windows=words[i:i+ window_size]
    window.append(windows) 
print(window)   
```

```python
training_pair=[]

for windows in window: 
    middle=len(windows)//2
    input_words=windows[middle]
    context_words=windows[:middle]+windows[middle+1:]
    for context_word in context_words:
        training_pair.append((input_words,context_word))

print(training_pair)
```

```python
vocab_size = len(vocab)
```

```python
def one_hot(words):
    vector = [0]* vocab_size
    index = word_to_index[words]
    vector[index]=1
    return vector

print(one_hot("queen"))
```

```python
X=[]
Y=[]
for input_word,output_word in training_pair:
    X.append(one_hot(input_word))
    Y.append(one_hot(output_word))
```
